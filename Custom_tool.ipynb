#Custom Tool

step 1: Create a function


```python
def multiply(a,b):
  "Multiply two numbers"
  return a*b
```

step 2: Add type hints

```python
def multiply(a: int, b:int) -> int:
  "Multiply two numbers"
  return a*b
```

step 3: Add a tool decorator

```python
from langchain_core.tools import tool

@tool
def multiply(a: int, b:int) -> int:
  "Multiply two numbers"
  return a*b
```

```python
results= multiply.invoke({"a":2, "b":3})
print(results)
```

# Method 2: Using Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class Mulitiply_input(BaseModel):
  a:int= Field(required= True, description= "first number")
  b:int= Field(required= True, decription= "Second number")

```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two numbers",
    args_schema=Mulitiply_input
)

results= multiply_tool.invoke({"a": 2, "b": 3})
```

# Base Tool


```python
from langchain_core.tools import BaseTool
from pydantic import BaseModel,Field
from typing import Type

class Multiply_input(BaseModel):
  a: int= Field(required= True, description= "First number")
  b: int= Field(required= True, description= "Second number")

class Multiply(BaseTool):
  name: str= "multiply"
  description :str= "Multiply two numbers"
  args_schema:Type[BaseModel]= Multiply_input

  def _run(self, a, b):
    return a*b

multi= Multiply()

result= multi.invoke({'a':2, 'b':7})
print(result)
```

# Toolkit


```python
from langchain_core.tools import tool

@tool
def multiply(a: int, b:int) -> int:
  "Multiply two numbers"
  return a*b

@tool
def add(a:int, b:int)->int:
  """ Addition of two numbers"""
  return a+b
```

```python
# toolkit

class MathsBasic:

  def get_tools(self):
    return[add, multiply]

toolkit= MathsBasic()
tools= toolkit.get_tools()

for i in tools:
  print(i.name, "=>", i.description)
```
